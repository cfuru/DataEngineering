```python
# Imports
from pyspark.sql.types import StringType
from Databricks.Notebooks.utils import pipeline_helpers
```

```python
# Define normalize_source_filename wrapper with error handling
def normalize_source_filename_impl(path):
    """
    Wrapper around pipeline_helpers.normalize_source_filename() with error handling.
    
    Args:
        path (str): Full file path (e.g., '/path/to/file.csv')
    
    Returns:
        str or None: Bare filename (e.g., 'file.csv') or None if invalid
    """
    try:
        return pipeline_helpers.normalize_source_filename(path)
    except Exception as e:
        print(f"Error normalizing filename for path '{path}': {e}")
        return None
```

```python
# Register normalize_source_filename as a Spark SQL UDF
spark.udf.register("normalize_source_filename", normalize_source_filename_impl, StringType())
print("✓ Registered SQL UDF: normalize_source_filename")
```

```python
# Runtime check: Verify UDF registration (optional - can fail gracefully outside Databricks)
try:
    result = spark.sql("SELECT normalize_source_filename('/some/path/to/file.csv') AS fname").collect()
    print(f"✓ UDF runtime check passed. Result: {result[0]['fname']}")
except Exception as e:
    print(f"⚠ UDF runtime check skipped (not in Databricks runtime): {e}")
```
